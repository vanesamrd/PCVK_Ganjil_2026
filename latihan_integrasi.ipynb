```python
import numpy as np
import pandas as pd
import cv2 as cv
from google.colab.patches import cv2_imshow
from skimage import io
from skimage import transform
from PIL import Image
import matplotlib.pyplot as plt
```

```python
urls = ["https://iiif.lib.ncsu.edu/iiif/0052574/full/800,/0/default.jpg",
        "https://iiif.lib.ncsu.edu/iiif/0016007/full/800,/0/default.jpg",
        "https://external-preview.redd.it/hJNL3HXwmJXzzIdyUKdsVQerrBIe0KbO7d26zEXWidE.jpg?auto=webp&s=4a8bf4911d40dddb64ebc30505e704003297f414"]

for url in urls:
    image = io.imread(url)
    image = cv.resize(image, (0,0), fx=0.5, fy=0.5)
    image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
    final_frame = cv.hconcat((image, image_2))
    cv2_imshow(final_frame)
    print('\n')
```

```python
tinggi = image_2.shape[0]
lebar = image_2.shape[1]
print("resolusi image: tinggi x lebar = ", tinggi, " x ", lebar)
cv2_imshow(image_2)
```

```python
image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_3 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

for y in range(lebar):
    image_3[int((tinggi)/2),y] = [255,255,255]

final_frame = cv.hconcat((image_2, image_3))
cv2_imshow(final_frame)
```

```python
#PRAKTIKUM D2
#menambahkan garis vertikal dan diagonal pada image
image_tugas = image_2.copy()

for y in range(tinggi):
    image_tugas[y, int(lebar / 2)] = [255, 255, 255]

cv2_imshow(image_tugas)

image_tugas = image_2.copy()

for y in range(tinggi):
    x = int(y * lebar / tinggi)
    image_tugas[y, x] = [255, 255, 255]

for y in range(tinggi):
    x = int(lebar - 1 - (y * lebar / tinggi))
    image_tugas[y, x] = [255, 255, 255]

cv2_imshow(image_tugas)
```

```python
#menambahkan garis horizontal berwarna putih di tengah gambar dengan panjang 200 pixel
image_tugas = image_2.copy()

y = int(tinggi / 2)

for x in range(100, 300):
    image_tugas[y, x] = [255, 255, 255]

cv2_imshow(image_tugas)
```

```python
#membuat kotak berwarna putih dengan ukuran 100 × 100 pixel
image_tugas = image_2.copy()

for y in range(50, 150):
    for x in range(50, 150):
        image_tugas[y, x] = [255, 255, 255]

cv2_imshow(image_tugas)
```
